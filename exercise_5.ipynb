program that counts the no.of vowels and consonants in a string

### Program to count vowels and consonants

```python
a=str(input("enter a string"))
b=a.lower()
vowels="aeiou"
count_v=0
count_c=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print("no.of vowels=",count_v)
print("no.of consonants=",count_c)
```
