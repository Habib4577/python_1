```python
#prime numbers in a given range
n = int(input())
m = int(input())

for i in range(n, m + 1):
    if i > 1:
        isPrime = 1
        for j in range(2, i):
            if i % j == 0:
                isPrime = 0
                break
        if isPrime == 1:
            print(i)
```

```python
#print sum of digits of a given number
n=int(input())
sum=0
while(n>0):
  r=n%10
  sum=sum+r
  n=n//10
print(sum)
```
