```python
#program for match
num=int(input("enter"))
match num:
  case 1:
    print("Sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("Saturday")
  case _:
    print("invalid input")
```
